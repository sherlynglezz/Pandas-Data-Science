```python
df_clean = df.dropna()
df_clean.shape
```

```python
df_clean.isnull().sum()
```

```python
df_filled = df.copy()
df_filled["Total day minutes"] = df_filled["Total day minutes"].fillna(
    df_filled["Total day minutes"].mean()
)
df_filled["Total day minutes"].isnull().sum()
```

```python
df_clean = df_clean.copy()
df_clean["Total minutes"] = (
    df_clean["Total day minutes"]
    + df_clean["Total eve minutes"]
    + df_clean["Total night minutes"]
    + df_clean["Total intl minutes"]
)

df_clean[["State", "Total minutes"]].head()
```

```python
df_clean.sort_values("Total minutes", ascending=False).head()
```

```python
df_without_total = df_clean.drop(columns=["Total minutes"])
df_without_total.shape
```

```python
df_clean["Day minutes in hours"] = df_clean["Total day minutes"].apply(
    lambda minutes: minutes / 60
)

df_clean[["Total day minutes", "Day minutes in hours"]].head()
```

```python
df_clean.to_csv("telco_churn_cleaned.csv", index=False)
```
